# STEP 5: FEATURE ENGINEERING 
When we say "feature engineering", features are our variables (ie. our columns) and engineering those means transforming them in a way such that we can derive patterns from the data. What that entails in the scope of machine learning is that the input variables for the model should be formatted in order for the model to understand them the way a human does, so we are converting our columns to useful features by doing that. 

```python
model_df = cleaned_df.copy()

model_df["is_senior"] = model_df["experience_level"].isin(["SE", "EX"]).astype(int)
model_df["log_salary"] = np.log(model_df["salary_in_usd"])
model_df["company_size_ord"] = model_df["company_size"].map({"S": 0, "M": 1, "L": 2})
top_locs = ["US", "GB", "CA", "ES", "DE", "IN", "FR"]
model_df["company_region"] = np.where(model_df["company_location"].isin(top_locs), model_df["company_location"], "Other")
model_df["works_abroad"] = (model_df["employee_residence"] != model_df["company_location"]).astype(int)
model_df["remote_type"] = model_df["remote_ratio"].map({0: "on-site", 50: "hybrid", 100: "remote"})

def get_job_family(title):
    t = title.lower()
    if "engineer" in t:
        return "Engineer"
    if "analyst" in t:
        return "Analyst"
    if "scientist" in t:
        return "Scientist"
    return "Other"

model_df["job_family"] = model_df["job_title"].apply(get_job_family)
print(model_df["job_family"].value_counts())

model_df[["job_title", "log_salary", "company_size_ord", "company_region",
          "works_abroad", "remote_type", "job_family", "is_senior"]].head()
```

Let's quickly visualize how the new features we created relate to the target variable (`is_senior`) to see if they might be useful for our model:

```python
plt.figure(figsize=(10, 5))
sns.countplot(
    data=model_df,
    x="job_family",
    hue=model_df["is_senior"].map({0: "entry/mid", 1: "senior/exec"}),
    order=model_df["job_family"].value_counts().index
)
plt.title("Seniority by Job Family")
plt.xlabel("Job Family")
plt.ylabel("Number of Roles")
plt.xticks(rotation=45)
plt.legend(title="Seniority", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```

# STEP 5.5: FEATURE SELECTION 
Here, our main goal is to just choose the columns that the model will be using. We want to avoid using columns that are overly specific like unique identifier columns, columns that are unavailable at the time of prediction, or columns that are too similar to the actual target. 

```python
target_col = "is_senior"

numeric_features = [
    "log_salary",
    "company_size_ord",
    "remote_ratio",
    "work_year",
    "works_abroad",
]
categorical_features = [
    "employment_type",
    "company_region",
    "remote_type",
    "job_family",
]

features = numeric_features + categorical_features

X = model_df[features]
y = model_df[target_col]

print("Feature columns:", features)
print("Target column:", target_col)
print("X shape:", X.shape)
print("y shape:", y.shape)

X.head()
```

# STEP 6: TRAIN / TEST SPLIT 
Our model needs to be trained on data in order to learn the patterns that exist within it and once we've created our model, we test it with examples of data that it hasn't seen before. In order to do this, we must split our data into "training" and "testing" data, where the training set teaches the model and the testing set estimates how the model will perform on real world / new data. # STEP 7: MODEL SELECTION 
We will break this step down a little further into the following: 
1. Preprocessing Pipeline 
2. Establishing a Baseline Model 
3. Model Selection and Training 

### Why does preprocessing matter? 

Raw data is almost never ready for a model to learn from, thus not ready for actual machine learning, which requires numbers. Real datasets often contain one or many of the following which make it unusable to the model: 

- missing values
- text categories
- columns on different scales 
- messy labels 
- irrelevant columns 
- dates 

Preprocessing is the step where we turn the real world data that you'd be given into a format that a machine can more easily read and understand. Here, you'll see we use a 'ColumnTransformer' because different column types need different treatments. 

### What is a pipeline and why does it matter? 

A pipeline essentially means chaining multiple steps together, instead of doing them manually. Here we can put the following into one object: 

1. cleaning data 
2. encoding categories 
3. scaling numbers 
4. training the model

## Preprocessing Pipeline 

We know that machine learning models rely on inputs being quantitative, thus this step of creating a preprocessing pipeline handles both categorial and numeric features. 

1. Numeric features --> fill any missing values left and scale them 
2. Categorical features --> fill any missing values and one-hot encode them 

Question: We know that models need numeric variables so turning categorical features into those is what we call one-hot encoding. Identify a few examples of what columns could be one-hot encoded and HOW to one-hot encode different types of data. 

ANSWER: A couple examples of what columns could be one-hot encoded, and how to one-hot encode different types of data is:
**Nominal categories (no order)** examples, employment_type (FT/PT/CT/FL), company_region, remote_type, job_family. each category becomes its own 0/1 column, e.g. job_family_Engineer = 1. here, we can use handle_unknown="ignore so a category that never appeared in training becomes all zeros instead of crashing.

Here, we want to fit preprocessing on the training data only, then apply it to the testing data after. 

**Important Key Terms to Know:** 

imputing: filling missing values with other values in the data (ie filling nan with mean/mode/median)

StandardScalar() --> preprocessing tool from Pipeline to scale numeric variables 

ColumnTransformer() --> does exactly what it sounds like, transforms columns 

OneHotEncoder() --> same thing, one-hot encodes columns 

```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

preprocessor
```

## Baseline Model

Before we train a real model, we build a simple baseline to answer: "How well could we do with a lazy strategy?" This one always predicts the most common class, which here is **senior/exec**.

From the train/test split we saw that ~64% of roles are senior/exec. So the baseline gets ~64% **accuracy** while learning nothing. That's why accuracy alone is misleading for this problem. We'll also track:

- **Balanced accuracy**: the average of how well we do on *each* class. Always guessing one class gives exactly 0.50, so anything above that is real signal. **This is our main metric.**
- **F1 for entry/mid (class 0)**: how well we catch the minority class.

```python
from sklearn.metrics import balanced_accuracy_score, f1_score

baseline_model = DummyClassifier(strategy="most_frequent")
baseline_model.fit(X_train, y_train)

baseline_preds = baseline_model.predict(X_test)

print("Baseline accuracy:         ", round(accuracy_score(y_test, baseline_preds), 3))
print("Baseline balanced accuracy:", round(balanced_accuracy_score(y_test, baseline_preds), 3))
print("Most common class in training data:", y_train.value_counts().idxmax(), "(1 = senior/exec)")
```

## Model Selection and Training

We train 6 models through the **same** preprocessing pipeline so the comparison is fair:

1. Logistic Regression: simple, interpretable, learns linear relationships
2. Decision Tree: rule-based splits, can capture non-linear patterns
3. Random Forest: an ensemble of many trees, usually stronger but less interpretable
4. Extra Trees: like Random Forest but with more randomness
5. Gradient Boosting: builds trees one after another, each fixing the last one's mistakes
6. Hist Gradient Boosting: a faster version of gradient boosting

### What is a class imbalance?

Our data is ~64% senior/exec and ~36% entry/mid. A lazy model can lean toward predicting "senior" and still look decent on accuracy, while missing most junior roles.

`class_weight="balanced"` tells a model to pay more attention to the less common class (entry/mid). Gradient Boosting doesn't support this option, so watch how it behaves compared to the others.

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=RANDOM_STATE,
        class_weight="balanced",
        max_depth=8,
        min_samples_leaf=5
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=500,
        random_state=RANDOM_STATE,
        class_weight="balanced",
        min_samples_leaf=2
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=500,
        random_state=RANDOM_STATE,
        class_weight="balanced",
        min_samples_leaf=2
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=RANDOM_STATE
    ),

    "Hist Gradient Boosting": HistGradientBoostingClassifier(
        class_weight="balanced",
        random_state=RANDOM_STATE
    )
}

results = []
trained_pipelines = {}

for name, model in models.items():
    pipe = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipe.fit(X_train, y_train)
    preds = pipe.predict(X_test)

    trained_pipelines[name] = pipe
    results.append({
        "model": name,
        "balanced_accuracy": balanced_accuracy_score(y_test, preds),
        "f1_entry_mid": f1_score(y_test, preds, pos_label=0),
        "accuracy": accuracy_score(y_test, preds)
    })

results_df = pd.DataFrame(results).sort_values("balanced_accuracy", ascending=False).round(3)
results_df
```

### Why compare multiple models?

No single model is best for every dataset, because each one learns patterns differently. Notice that **Gradient Boosting usually has the highest plain accuracy but not the highest balanced accuracy**. Without `class_weight`, it leans toward predicting "senior". It looks good on accuracy but is worse at catching entry/mid roles, which is exactly the trap balanced accuracy protects us from.

All the models land fairly close together. That tells us the **features** (mainly salary) limit how well we can do, not the choice of algorithm.

```python
best_model_name = results_df.iloc[0]["model"]
best_model = trained_pipelines[best_model_name]

print("Best model (by test balanced accuracy):", best_model_name)
```

# STEP 8: MODEL EVALUATION
This is where we see how well the model performs and where exactly it is failing. A single score is a good summary, but the classification report and confusion matrix show performance **per class**:

- **precision**: of the roles we *predicted* as this class, how many really were?
- **recall**: of the roles that *really* are this class, how many did we catch?

```python
class_names = ["entry/mid", "senior/exec"]
test_preds = best_model.predict(X_test)

print("Test balanced accuracy:", round(balanced_accuracy_score(y_test, test_preds), 3))
print("Test accuracy:         ", round(accuracy_score(y_test, test_preds), 3))
print()
print("Classification report:")
print(classification_report(y_test, test_preds, target_names=class_names))
```

```python
cm = confusion_matrix(y_test, test_preds)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_names, yticklabels=class_names)
plt.title(f"Confusion Matrix: {best_model_name}")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()
```

**How to read it:** the diagonal (top-left, bottom-right) holds correct predictions. The top-right box counts entry/mid roles we wrongly called senior, and the bottom-left box counts senior roles we wrongly called entry/mid.

## Cross-Validation

Cross-validation gives a more reliable estimate by training and testing the model several times on different splits of the data.

5-fold cross-validation:
1. split the data into 5 parts
2. train on 4 parts
3. test on the remaining part
4. repeat 5 times
5. average the scores

### Why one train/test split isn't enough:

One split can be lucky or unlucky. Cross-validation evaluates each model 5 times and averages, which gives a more stable comparison.

**Important:** we run CV on the **training set only** (`X_train`). The test set stays locked away until the very end, so our final test score stays honest.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

cv_results = []

for name, model in models.items():
    pipe = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring="balanced_accuracy", n_jobs=-1)

    cv_results.append({
        "model": name,
        "mean_cv_balanced_accuracy": scores.mean(),
        "std_cv_balanced_accuracy": scores.std()
    })

cv_results_df = pd.DataFrame(cv_results).sort_values("mean_cv_balanced_accuracy", ascending=False).round(3)
cv_results_df
```

**How to read this:** a higher mean is better, and a smaller std means more stable results across folds. If two models are within about one std of each other, they're basically tied, so prefer the simpler one.

Here the top models are within ~0.01–0.02 of each other, which is about the size of the std. No model clearly dominates.

# Hyperparameter Tuning

A model has two types of settings:
1. **Parameters** are learned from the data, e.g. the split points inside a decision tree.
2. **Hyperparameters** are settings we choose before training. Grid search tries every combination and uses cross-validation to pick the best one.

Examples:
- number of trees
- max tree depth
- minimum samples per leaf
- max number of features considered at each split

We tune a **Random Forest**. It's one of the stronger models here, it's stable, and it gives us feature importances to interpret afterward. We tune for **balanced accuracy**, the same metric we've used everywhere else.

```python
tuning_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1))
])

param_grid = {
    "model__n_estimators": [200, 500],
    "model__max_depth": [None, 6, 12],
    "model__min_samples_leaf": [1, 3, 8],
    "model__max_features": ["sqrt", None]
}

grid_search = GridSearchCV(
    tuning_pipeline,
    param_grid=param_grid,
    cv=cv,
    scoring="balanced_accuracy",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best CV balanced accuracy:", round(grid_search.best_score_, 3))

final_model = grid_search.best_estimator_
```

# STEP 9: FINAL MODEL EVALUATION
Now we unlock the test set one last time to evaluate the tuned model.

```python
final_preds = final_model.predict(X_test)

print("Final tuned model on the test set:")
print("  balanced accuracy:", round(balanced_accuracy_score(y_test, final_preds), 3), "(baseline = 0.5)")
print("  F1 (entry/mid):   ", round(f1_score(y_test, final_preds, pos_label=0), 3))
print("  accuracy:         ", round(accuracy_score(y_test, final_preds), 3), "(baseline =", round(accuracy_score(y_test, baseline_preds), 3), ")")
print()
print("Final tuned model classification report:")
print(classification_report(y_test, final_preds, target_names=class_names))
```

## Model Interpretability

Accuracy tells us how well the model performs, and feature importance helps us understand *why*.

### 1. Built-in feature importance
Random Forest scores each feature by how much it helped the trees split. It's quick, but it tends to over-credit numeric columns like salary.

```python
feature_names = final_model.named_steps["preprocessor"].get_feature_names_out()
importances = final_model.named_steps["model"].feature_importances_

importance_df = pd.DataFrame({
    "feature": feature_names,
    "importance": importances
}).sort_values("importance", ascending=False)

plt.figure(figsize=(8, 6))
sns.barplot(data=importance_df.head(15), x="importance", y="feature")
plt.title("Top 15 Features (Random Forest Importance)")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()

importance_df.head(15)
```

### 2. Permutation importance (more trustworthy)
Shuffle one **original** column at a time on the test set and measure how much balanced accuracy drops. If scrambling a column hurts the model a lot, the model really relies on it.

```python
from sklearn.inspection import permutation_importance

perm = permutation_importance(
    final_model, X_test, y_test,
    scoring="balanced_accuracy",
    n_repeats=15,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

perm_df = pd.DataFrame({
    "feature": X_test.columns,
    "drop_in_balanced_accuracy": perm.importances_mean
}).sort_values("drop_in_balanced_accuracy", ascending=False)

plt.figure(figsize=(8, 4))
sns.barplot(data=perm_df, x="drop_in_balanced_accuracy", y="feature")
plt.title("Permutation Importance")
plt.xlabel("Drop in Balanced Accuracy When Shuffled")
plt.ylabel("Feature")
plt.show()

perm_df.round(4)
```

## Error Analysis

Accuracy tells us how often the model is correct, while error analysis tells us *where* it goes wrong. This is the most important real-world DSML skill, because you can only improve a model once you know where it's making mistakes!

We bring back the original `experience_level`, `job_title`, and `salary_in_usd` so the mistakes are easy to read.

```python
errors = X_test.copy()
errors["actual"] = y_test.values
errors["predicted"] = final_preds
errors["experience_level"] = model_df.loc[X_test.index, "experience_level"]
errors["job_title"] = model_df.loc[X_test.index, "job_title"]
errors["salary_in_usd"] = model_df.loc[X_test.index, "salary_in_usd"]
errors["is_wrong"] = errors["actual"] != errors["predicted"]

print("Misclassified:", errors["is_wrong"].sum(), "of", len(errors))
print()
print("Error rate by original experience level:")
print(errors.groupby("experience_level")["is_wrong"].mean().round(3))
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=errors,
    x="experience_level",
    y="salary_in_usd",
    hue=errors["is_wrong"].map({False: "correct", True: "wrong"}),
    order=["EN", "MI", "SE", "EX"]
)
plt.yscale("log")
plt.title("Salary of Correct vs. Misclassified Roles")
plt.xlabel("Experience Level")
plt.ylabel("Salary (USD, log scale)")
plt.legend(title="Prediction")
plt.show()
```

```python
# the actual mistakes, lowest salary first
errors[errors["is_wrong"]][
    ["job_title", "experience_level", "salary_in_usd", "company_region", "remote_type", "job_family", "predicted"]
].sort_values("salary_in_usd").head(15)
```

**What the errors tell us:**
- **Mid-level (MI) is the hardest group.** It has the highest error rate because MI salaries overlap with both entry-level and senior salaries.
- In the boxplot, the **misclassified seniors are the low earners** and the **misclassified entry/mid roles are the high earners**. The model basically draws a salary line, and anyone on the "wrong" side of it gets misclassified.
- This confirms that salary is doing almost all the work, which matches the permutation importance.
